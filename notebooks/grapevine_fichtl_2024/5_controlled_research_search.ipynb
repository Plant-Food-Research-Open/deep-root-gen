# Controlled research search

This notebook combines the upstream Optuna workflow with the current controlled
root generator and versioned comparison. It uses invented examples by default.
No empirical search or fresh-seed evaluation runs until explicitly enabled.
See [method and input contracts](../../docs/research/NOTEBOOK_PARAMETER_SEARCH.md).

The bridge supports maximum order 3, 4 or 5 with explicit higher-order assumptions
and separately versioned 13/15/17-trait diagnostics. All roots enter profiles.
The optional local D-04 case conditions on exported class counts/lengths; its
raw parent IDs confirm those classes, while RSML XML nesting is inconsistent.
Measured attachment edges are excluded from the length operator. Its SRL-derived biomass reference remains outside the objective.


```python
from pathlib import Path
import json
import subprocess
import sys

REPO = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "deeprootgen").is_dir())
CASE = "synthetic"  # Change explicitly to "local_d04_order5" for the private starter.
CASES = {
    "synthetic": ("examples/research/search_synthetic.json", ".runtime/notebook-search-synthetic"),
    "local_d04_order5": (
        ".runtime/research/so4-t4-order5-20261115-v1/d04_bundle/search_config.private.json",
        ".runtime/notebook-search-d04-order5-v1",
    ),
}
CONFIG, OUTPUT = (REPO / relative for relative in CASES[CASE])
RUN_SEARCH = False
RUN_QUALIFICATION = False
RESUME = False
SCRIPT = REPO / "scripts/research/parameter_search.py"
```

## Validate the explicit route and local inputs

The configuration chooses `population_mean` or `per_seed_mean`, fixed development
and disjoint qualification seeds, a trial/startup budget and an output admission
threshold. Validation creates no search directory. Parameter and reference file
paths are resolved relative to the configuration file. Keep measured inputs local.


```python
command = [sys.executable, str(SCRIPT), "--config", str(CONFIG), "--outdir", str(OUTPUT)]
subprocess.run(command, cwd=REPO, check=True)
```

```python
if RUN_SEARCH:
    search_command = command + ["--execute"] + (["--resume"] if RESUME else [])
    subprocess.run(search_command, cwd=REPO, check=True)
    summary = json.loads((OUTPUT / "search/summary.json").read_text())
    display(summary)
else:
    print("Search not executed. Set RUN_SEARCH=True after reviewing the configuration.")
```

## Evaluate the frozen candidate on fresh seeds

This verifies the selected trial against its durable search record, then uses
only the qualification seeds declared before search. It does not optimize again.
Fresh seeds assess Monte Carlo robustness, not independent biological accuracy.


```python
if RUN_QUALIFICATION:
    subprocess.run(command + ["--phase", "qualify", "--execute", "--resume"], cwd=REPO, check=True)
    qualification = json.loads((OUTPUT / "qualification.json").read_text())
    display(qualification)
else:
    print("Fresh-seed evaluation not executed.")
```
